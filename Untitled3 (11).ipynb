```python
def descargar_csv_con_cluster(df_original, df_clusterizado, labels, nombre_algoritmo, file_name):
    resultado = df_original.copy()
    resultado["cluster"] = np.nan
    resultado.loc[df_clusterizado.index, "cluster"] = labels

    nombre_salida = f"{file_name.rsplit('.', 1)[0]}_{nombre_algoritmo}_clusters.csv"
    resultado.to_csv(nombre_salida, index=False)
    files.download(nombre_salida)

    print(f"CSV generado y descargado: {nombre_salida}")
    return resultado
```

```python
def separar_columnas(df):
    ID_COL = df.columns[0]

    DESC_COLS = []
    FEATURE_COLS = []

    for col in df.columns[1:]:
        if pd.api.types.is_numeric_dtype(df[col]):
            if df[col].nunique(dropna=True) > 1:
                FEATURE_COLS.append(col)
            else:
                DESC_COLS.append(col)
        else:
            DESC_COLS.append(col)

    return ID_COL, DESC_COLS, FEATURE_COLS
```

```python
def elegir_k_elbow(X_scaled, k_min=2, k_max=10, random_state=42):
    ks = list(range(k_min, k_max + 1))
    inertias = []

    for k in ks:
        kmeans = KMeans(n_clusters=k, n_init=20, random_state=random_state)
        kmeans.fit(X_scaled)
        inertias.append(kmeans.inertia_)

    # segunda derivada aproximada
    dif1 = np.diff(inertias)
    dif2 = np.diff(dif1)

    best_idx = np.argmax(np.abs(dif2)) + 2
    best_k = ks[best_idx]

    return best_k, ks, inertias
```

```python
def elegir_eps_dbscan(X_scaled, min_samples=20):
    # Distancia al vecino número min_samples
    neighbors = NearestNeighbors(n_neighbors=min_samples)
    neighbors_fit = neighbors.fit(X_scaled)
    distances, _ = neighbors_fit.kneighbors(X_scaled)

    # Tomamos la última columna: distancia al vecino k
    k_distances = np.sort(distances[:, -1])

    # Coordenadas de la curva
    n = len(k_distances)
    x = np.arange(n)
    y = k_distances

    # Recta entre el primer y último punto
    p1 = np.array([x[0], y[0]])
    p2 = np.array([x[-1], y[-1]])

    # Distancia perpendicular de cada punto a la recta
    line_vec = p2 - p1
    line_vec = line_vec / np.linalg.norm(line_vec)

    point_vecs = np.column_stack((x, y)) - p1
    projections = np.outer(np.dot(point_vecs, line_vec), line_vec)
    orthogonal_vecs = point_vecs - projections
    dist_to_line = np.linalg.norm(orthogonal_vecs, axis=1)

    best_idx = np.argmax(dist_to_line)
    eps_value = k_distances[best_idx]

    return eps_value, k_distances, best_idx
```

```python
def reasignar_clusters_pequenos(X_scaled, labels_db, min_cluster_size=9000, incluir_ruido=True):
    labels_db = np.array(labels_db).copy()

    # contar tamaño de cada cluster
    unique, counts = np.unique(labels_db, return_counts=True)
    cluster_sizes = dict(zip(unique, counts))

    # clusters grandes: tamaño suficiente y no ruido
    large_clusters = [
        c for c, size in cluster_sizes.items()
        if size >= min_cluster_size and c != -1
    ]

    if len(large_clusters) == 0:
        print("No hay clusters grandes para reasignar.")
        return labels_db

    # dataframe temporal
    X_df = pd.DataFrame(X_scaled)
    X_df["cluster"] = labels_db

    # centroides de clusters grandes
    centroids = (
        X_df[X_df["cluster"].isin(large_clusters)]
        .groupby("cluster")
        .mean()
    )

    # reasignar puntos de clusters pequeños y opcionalmente ruido
    for i, label in enumerate(labels_db):
        condicion = (label == -1 and incluir_ruido) or (
            label != -1 and cluster_sizes[label] < min_cluster_size
        )

        if condicion:
            closest = pairwise_distances_argmin(
                X_scaled[i].reshape(1, -1),
                centroids.values
            )[0]

            labels_db[i] = centroids.index[closest]

    return labels_db
```

```python
def renumerar_clusters(labels):
    labels = np.array(labels)
    unique_labels = sorted(np.unique(labels))
    mapping = {old: new for new, old in enumerate(unique_labels)}
    return np.array([mapping[x] for x in labels])
```

```python
def elegir_eps_percentil(X, min_samples, percentil):
    nbrs = NearestNeighbors(n_neighbors=min_samples).fit(X)
    distances, _ = nbrs.kneighbors(X)

    k_distances = distances[:, -1]

    eps_value = np.percentile(k_distances, percentil)

    return eps_value, np.sort(k_distances)
```

```python


def samplear(X, frac=0.3, max_samples=20000, random_state=42):
    n = len(X)
    size = min(int(n * frac), max_samples)
    rng = np.random.default_rng(random_state)
    idx = rng.choice(n, size=size, replace=False)

    if isinstance(X, pd.DataFrame):
        return X.iloc[idx]
    else:
        return X[idx]
```

```python
def evaluar_hdbscan_fast(X, min_cluster_size, min_samples):
    import hdbscan
    import numpy as np

    clusterer = hdbscan.HDBSCAN(
        min_cluster_size=min_cluster_size,
        min_samples=min_samples
    )

    labels = clusterer.fit_predict(X)

    mask = labels != -1
    labels_clean = labels[mask]

    noise_ratio = 1 - mask.mean()

    if noise_ratio > 0.6:
        return -np.inf

    if len(labels_clean) == 0:
        return -np.inf

    unique, counts = np.unique(labels_clean, return_counts=True)

    n_clusters = len(unique)
    if n_clusters < 2 or n_clusters > 10:
        return -np.inf

    largest_ratio = counts.max() / counts.sum()
    if largest_ratio > 0.8:
        return -np.inf

    # equilibrio (rápido)
    p = counts / counts.sum()
    entropy = -np.sum(p * np.log(p + 1e-12))
    cluster_penalty = max(0, n_clusters - 8) * 0.2

    score = entropy - 1.5 * noise_ratio - 1.0 * largest_ratio - cluster_penalty

    return score
```

```python
def buscar_mejor_hdbscan_fast(X):

    X_sample = samplear(X, frac=0.3)

    min_cluster_sizes = [200, 500, 1000, 2000, 4000, 8000]
    min_samples_values = [5, 15, 30, 50]

    best_score = -np.inf
    best_params = None

    for mcs in min_cluster_sizes:
        for ms in min_samples_values:
            if ms > mcs:
                continue

            score = evaluar_hdbscan_fast(X_sample, mcs, ms)

            if score > best_score:
                best_score = score
                best_params = (mcs, ms)

    return best_params
```
