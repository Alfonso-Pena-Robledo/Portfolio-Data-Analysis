```python
# Variables RFM
X = np.log1p(rfm_basico[['recency','frequency','monetary']])

# Escalar
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Método del codo
inertia = []
K = range(1,11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Gráfico
plt.plot(K, inertia, marker='o')
plt.xlabel("Número de clusters")
plt.ylabel("Inertia")
plt.title("Método del codo")
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)

labels = kmeans.fit_predict(X_scaled)
rfm_basico['cluster'] = labels
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data=rfm_basico, x='frequency', y='monetary', hue='cluster', palette='Set2')
plt.show()
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# DataFrame con componentes y clusters
pca_rfm_basico = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_rfm_basico["cluster_kmeans"] = labels
```

```python
plt.figure(figsize=(8, 6))

for cluster_id in sorted(pca_rfm_basico["cluster_kmeans"].unique()):
    subset = pca_rfm_basico[pca_rfm_basico["cluster_kmeans"] == cluster_id]
    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=f"Cluster {cluster_id}",
        alpha=0.5,
        s=15
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Visualización PCA de clusters (K-means)")
plt.legend()
plt.show()
```

```python
rfm_scores = rfm_basico.copy()

# Recency (invertido: menor = mejor)
rfm_scores['R_score'] = pd.qcut(rfm_scores['recency'], 5, labels=[5,4,3,2,1])

# Frequency (mayor = mejor)
rfm_scores['F_score'] = pd.qcut(rfm_scores['frequency'].rank(method='first'), 5, labels=[1,2,3,4,5])

# Monetary (mayor = mejor)
rfm_scores['M_score'] = pd.qcut(rfm_scores['monetary'], 5, labels=[1,2,3,4,5])
```

```python
rfm_scores['RFM_score'] = (
    rfm_scores['R_score'].astype(int) +
    rfm_scores['F_score'].astype(int) +
    rfm_scores['M_score'].astype(int)
)
```

```python
def segment_rfm(row):

    if row['R_score'] >= 4 and row['F_score'] >= 4 and row['M_score'] >= 4:
        return "Champions"

    elif row['F_score'] >= 4 and row['M_score'] >= 4:
        return "Loyal"

    elif row['R_score'] >= 4:
        return "Recent"

    elif row['R_score'] <= 2 and row['F_score'] >= 3:
        return "At Risk"

    else:
        return "Lost"


rfm_scores['segment'] = rfm_scores.apply(segment_rfm, axis=1)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    tabla_prop * 100,
    annot=True,
    fmt=".1f",
    cmap="magma"
)

plt.title("Distribución porcentual de segmentos RFM por cluster")
plt.ylabel("Segmento RFM")
plt.xlabel("Cluster")

plt.show()
```
